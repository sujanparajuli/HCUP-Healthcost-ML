```python
import os
import pandas as pd
```

```python
csv_path = os.path.expanduser("~/Documents/Research/diagnosis_data.csv")
ccsr_path = os.path.expanduser("~/Documents/Research/ccsr.xlsx")
output_path = os.path.expanduser("~/Documents/Research/diagnosis_data_with_ccsr.csv")
```

```python
ccsr_df = pd.read_excel(ccsr_path, usecols=["ICD-10-CM Code", "CCSR Category"])

code_to_ccsr = dict(zip(ccsr_df["ICD-10-CM Code"], ccsr_df["CCSR Category"]))
```

```python
chunk_size = 100000
chunks = pd.read_csv(csv_path, chunksize=chunk_size)


header_written = False
```

```python
for chunk in chunks:
    # Map the K088 column to CCSR Category
    chunk["CCSR Category"] = chunk["K088"].map(code_to_ccsr)

    # Append to output file
    if not header_written:
        chunk.to_csv(output_path, index=False, mode='w')
        header_written = True
    else:
        chunk.to_csv(output_path, index=False, mode='a', header=False)

print("Finished processing and saved to:", output_path)
```
